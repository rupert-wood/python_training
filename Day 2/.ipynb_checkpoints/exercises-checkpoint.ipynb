```python
def checkage(age):
    if age >=18:
        print("Cat A")
    elif age >=16:
        print("Cat B")
    else:
        print("Cat C")
checkage(13)
```

```python
def find_squares_up_to_n(n):
    x = 1
    while (x**2) < n:
        print(f'{x=}, x^2={x**2}')
        x += 1
find_squares_up_to_n(2000)
```

```python
def time_to_target(initial, target, rate):
    periods = 0
    value = initial
    while(value < target):
        periods += 1
        value = value * (1 + rate/100)
    return periods

time_to_target(100, 1000, 10)
```
